#📍 Imputing Missing Values

#### ⭐ For categorical features using 'mode' & for numerical features using 'median'

```python
# Age
df.Age.fillna(df.Age.median(), inplace=True)

# TypeofContract
df.TypeofContact.fillna(df.TypeofContact.mode()[0], inplace=True)

# DurationOfPitch
df.DurationOfPitch.fillna(df.DurationOfPitch.median(), inplace=True)

# NumberOfFolLowups
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode()[0], inplace=True)

# PreferredPropertyStar
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode()[0], inplace=True)

# NumberOfTrips
df.NumberOfTrips.fillna(df.NumberOfTrips.median(), inplace=True)

#NumberOfChildrenVisiting
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode()[0], inplace=True)

# MonthlyIncome
df.MonthlyIncome.fillna(df.MonthlyIncome.mean(), inplace=True)
```

# 📍 Feature Type checking

```python
# Categorical features
cat_cols = df.select_dtypes(include=['object']).columns
print('Total number of Categorical features in the dataset:',len(cat_cols))
```

```python
# Numerical features
numeric_cols = df.select_dtypes(include=['number']).columns
print('Total number of Numerical features in the dataset:',len(numeric_cols))
```

```python
# Discrete features
discrete_cols = [col for col in numeric_cols if len(df[col].unique()) < 25]
print('Total number of Discrete features in the dataset:',len(discrete_cols))
```

```python
# Continuous features
cont_cols = [col for col in numeric_cols if col not in discrete_cols]
print('Total number of Discrete features in the dataset:',len(cont_cols))
```
