# 🎯 Handling Unknown Values

###OrdinalEncoder (categories=[ ], handle_unknown = 'use_encoded_value' , unkonwn_value = ... )

```python
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']],
                    handle_unknown='use_encoded_value',
                    unknown_value=-1)
X_train, X_test, y_train, y_test = train_test_split(
    df.iloc[:,0:2], df.iloc[:,-1], test_size=0.2)
X_train_2 = oe.fit_transform(X_train)
oe.transform(np.array(['Poor','college']).reshape(1,2))
# If multiple unknown categories appear, OrdinalEncoder
# assigns the same unknown_value to all of them, because it
# treats them as a single unseen category group.
```

# 🎯 Handling Infrequent Categories

###OrdinalEncoder (max_categories = ...)
###OrdinalEncoder (min_fequency = ...)

```python
# handling infrequent categories
X = np.array([['dog'] * 5 + ['cat'] * 20 + ['rabbit'] * 10 +['snake'] * 3 + ['horse'] * 2], dtype=object).T
X
```

Approach 1 : Explicitly Declaring no. of Categories

```python
enc = OrdinalEncoder(max_categories=3).fit(X)
```

```python
enc.categories_
```

```python
enc.infrequent_categories_
```

```python
enc.transform(np.array([['cat','rabbit','snake','dog','horse']]).reshape(5,1))
```

Approach 2 : Frequency Count

```python
enc = OrdinalEncoder(min_frequency=5).fit(X)
```

```python
enc.infrequent_categories_
```

```python
enc.transform(np.array([['cat','dog','rabbit','snake','horse']]).reshape(5,1))
```

#🎯 Handling Missing Values

### OrdinalEncoder (encoded_missing_value=-1)

```python
# Example categorical data with missing values
data = [['Cat'], [np.nan], ['Dog'], ['Fish'], [np.nan]]
encoder = OrdinalEncoder(encoded_missing_value=-1)
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```
