# 🎯 Multi Label Classification

```python
from sklearn.preprocessing import MultiLabelBinarizer
data = [
    ['Action','Comedy'],
    ['Comedy', 'Drama'],
    ['Action','Drama']
]
mlb = MultiLabelBinarizer()
encoded = mlb.fit_transform(data)
print(encoded)
```

```python
mlb.inverse_transform(encoded)
```
