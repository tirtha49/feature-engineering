```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False)
```

```python
df = pd.DataFrame({
    'Gender': ['Male','Female','Female','Male'],
    'Education': ['UG','PG','School','UG'],
    'City': ['Delhi','Mumbai','Delhi','Kolkata']
})
```

```python
df
```

```python
ohe.fit_transform(df[['Gender','City']])
```

```python
# Which feature is divided into 0/1s
ohe.feature_names_in_
```

```python
# All sub-features of the main-feature
ohe.categories_
```

```python
# All sub-features_main-feature present
ohe.get_feature_names_out()
```

```python
# Entire 0/1 Matrix
pd.DataFrame(ohe.fit_transform(df[['Gender','City']]),columns=ohe.get_feature_names_out())
```

```python
# 0/1s to categorical
ohe.inverse_transform([[0,1,0,1,0]])
```

# 🎯 Handling Unknown

```python
# ohe = OneHotEncoder(..., handle_unknown = 'ignore', ...)
```

```python
df = pd.DataFrame({
    'Gender': ['Male','Female','Female','Male'],
    'Education': ['UG','PG','School','UG'],
    'City': ['Delhi','Mumbai','Delhi','Kolkata']
})
```

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown ='ignore')
```

```python
# Entire 0/1 Matrix
pd.DataFrame(ohe.fit_transform(df[['Gender','City']]),columns=ohe.get_feature_names_out())
```

```python
ohe.inverse_transform(np.array([0,1,1,0,0]).reshape(1,5))
```

```python
ohe.transform(np.array([['Male','Delhi']]))
```

```python
#‘ignore’ : When an unknown category is encountered
# one-hot encoded columns for this feature will be all zeros
ohe.transform(np.array([['DLFJOERJGROG', 'Kolkata']]))
```
